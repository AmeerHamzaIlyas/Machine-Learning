Classification Metrics models

Models Are
1.Accuracy_score
2.Precision_score
3.Recall_score 
4.f1_score (combination of precision and recall)
5.Confusion_matrix

```python
from sklearn.metrics import(
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
```

Confusion Matrix
    [[TN,FP]
    [FN,TP]]

TN -> True Negative
FP -> False Positive
FN -> False Negative
TP -> True Positive

```python
y_actual = [1,1,0,1,0,1,1,0,1]
y_predic = [1,0,0,1,0,1,0,1,0]

print("accuracy_score",accuracy_score(y_actual,y_predic))

print("precision_score",precision_score(y_actual,y_predic))

print("recall_score",recall_score(y_actual,y_predic))

print("f1_scor",f1_score(y_actual,y_predic))

print("confusion_matrix",confusion_matrix(y_actual,y_predic))

```

confusion_matrix [[2 1]
 [3 3]]

Here cofusion matrix means:

TN -> 2 True Negative
FP -> 1 False Positive
FN -> 3 False Negative
TP -> 3 True Positive
