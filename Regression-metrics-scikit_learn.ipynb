```python
# import scikit-learn matrices
from sklearn.metrics import(
    mean_absolute_error, # MAE
    mean_squared_error, # MSE
    root_mean_squared_error, # RMSE
    r2_score # R2 (Coefficient of determination)
)
```

```python
# using metrics models
y_actual = [100,150,200]
y_predic = [110,140,215]
# MAE
print("MAE ->",mean_absolute_error(y_actual,y_predic)) 

# MSE
print("MSE ->",mean_squared_error(y_actual,y_predic))

# RMSE
print("RMSE ->",root_mean_squared_error(y_actual,y_predic)) 
 
 # R2
print("R2 ->",r2_score(y_actual,y_predic)) 
 
```
