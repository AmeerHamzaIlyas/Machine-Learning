```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#simple linear regression
from sklearn.linear_model import LinearRegression
```

```python
#load data from seaborn
df = sns.load_dataset("tips")
```

```python
# create a regression line plot using seaborn
sns.lmplot(x='total_bill',y='tip',data=df)
```

```python
#select the feature/input column from df
X = df[['total_bill']] # double squre brackets means data in 2D (x, y)
#select the target/output column from df
y = df[['tip']]
```

```python
#Load/Call the linear reggression  model
model = LinearRegression()

# Fit the models means we train the model
model.fit(X,y)
```

```python
# lets predict the tip of unknown total_bill
# total_bill = 50
model.predict([[50]])
```

```python
# whole logic to train a ml model 

#import libraries
import seaborn as sns
from sklearn.linear_model import LinearRegression
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill"]]
# target/output
y=df[["tip"]]
#used the model
model = LinearRegression()
#train the model
model.fit(X,y)
#predict the outcomes/results
model.predict([[50]])
```
