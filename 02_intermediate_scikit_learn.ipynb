```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#simple linear regression
from sklearn.linear_model import LinearRegression
df = sns.load_dataset("tips")
```

1. simple linear regression model

```python
# lmplot means linear model plot
sns.lmplot(x="total_bill", y="tip",data=df)
```

```python
# whole logic to train a ml model 

#import libraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill"]]
# target/output
y=df[["tip"]]
#used the model
model = LinearRegression()
#train the model
model.fit(X,y)
#predict the outcomes/results
model.predict([[50]])
```

# multiple linear regression

```python
# training model by using two inputs lables/features

#import libraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill","size"]]
# target/output
y=df[["tip"]]
#used the model
model = LinearRegression()
#train the model
model.fit(X,y)
#predict the outcomes/results
model.predict([[50,3]])
```

#convert categoric form data to numeric one that is nessesary

```python
#import libraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
#load datasets
sns.load_dataset("tips")
df.head()
# creating changing in new added column rather then existing-one
df["smoker_num"] = df["smoker"].map({'Yes':1,'No':0}) # manually possinble by pandas
df.head()
```

```python
# training model by using multiple inputs lables/features

#import libraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill","size","smoker_num"]]
# target/output
y=df[["tip"]]
#used the model
model = LinearRegression()
#train the model
model.fit(X,y)
#predict the outcomes/results
print(model.predict([[200,10,1]])) # for smokers
print(model.predict([[200,10,0]])) # for non-smokers
```

1.logistic Regression (it is not regression but classification means(yes 0r no))
Logistic regression is also called Binary regression.
Logistic regression is also called Sigmoid Function.

```python
#import libraries
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill","size","tip"]]
# target/output
y=df[["smoker"]]
#used the model
model = LogisticRegression()
#train the model
model.fit(X,y)
#predict the outcomes/results
model.predict([[400,3,4]])
```

Decision Tree Classifier Algorithm

```python
#import libraries
import pandas as pd
import seaborn as sns
#import naive bayes
from sklearn.naive_bayes import GaussianNB
#import Decision Tree
from sklearn.tree import DecisionTreeClassifier
#load datasets
sns.load_dataset("tips")
# feature/input
X=df[["total_bill","size","tip"]]
# target/output
y=df[["smoker"]]
#used the model
model = DecisionTreeClassifier()
#train the model
model.fit(X,y)
#predict the outcomes/results
model.predict([[400,3,4]])
```
