1.Imputing values as mode, median, mean etc
for Regression --> mean,median
for Clssification -->mode

```python
# impute missing values with mean/median etc
data['age'] = data['age'].fillna(data['age'].mean())
data.isnull().sum().sort_values(ascending=False)
```

```python
# impute missing values with mode etc
data['embark_town'] = data['embark_town'].fillna(data['embark_town'].mode()[0])
data['embarked'] = data['embarked'].fillna(data['embarked'].mode()[0])
data.isnull().sum().sort_values(ascending=False)
```

Converting Categorical data columns to numerical one using automatical methods like LabelEncoder

```python
#we take all str and category columns for labelen-Coding
from sklearn.preprocessing import LabelEncoder

#columns to encode
columns_to_encode = ['sex', 'embarked', 'who', 'deck', 'class', 'embark_town', 'alive']

# 3dictionary to store labelencoder for each column
label_encoders = {}

#loop to apply labelencoder for the columns
for col in columns_to_encode:
    # call labelencoder object
    le = LabelEncoder()
    # train and transform
    df[col] = le.fit_transform(df[col])
    # add in the dictionary one-one
    label_encoders[col] = le
    
df.head()
```

Imputing the missing values by iterative-imputer


```python
imputer  = IterativeImputer(max_iter=10) #max-iterator means who many times algo check the values by iteration
#columns to impute
columns_to_impute = ['age', 'embark_town', 'embarked', 'deck']

#loop
for col in columns_to_impute:
    df[col] = imputer.fit_transform(df[[col]])
# check missing values
df.isnull().sum().sort_values(ascending=False) 
```

```python
#Inverse transform for encoded  columns
for col in columns_to_encode:
    #retreive the corresponding lableencoder for the column
    le = label_encoders[col]
    #inverse transform the data and convert to integer type
    df[col] = le.inverse_transform(df[col].astype(int))
    
df.head()
```

Random Forest for Imputing Missing Values
bagging , Bosting, Esamble

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error, r2_score
from sklearn.impute import SimpleImputer

df = sns.load_dataset('titanic')

df.isnull().sum().sort_values(ascending=False)
```

```python
# Remove deck column
df.drop('deck',axis=1,inplace=True)

df.isnull().sum().sort_values(ascending=False)
```

```python
df.head()
```

```python
df.info()
```

```python
# Label to encode (encode means convert the dtr,category type to integer or  other)
from sklearn.preprocessing import LabelEncoder
# columns to encoding required
columns_to_encode = ['sex', 'embarked', 'class', 'who', 'embark_town', 'alive']

# Dictionary for label encoders
label_encoders = {}

#loop to iterate all values
for col in columns_to_encode:
    #call lable encoder and store in variable then use that variable
    le = LabelEncoder()
    # Fit and transform
    df[col] = le.fit_transform(df[col])
    #store in dictionary
    label_encoders[col] = le
df.head()
```

```python
# split the dataset into two parts : one with missing values and other is non-missing values
df_with_missing = df[df['age'].isna()]
#drop all rows with missing values
df_without_missing = df.dropna()
```

```python
print("original dataset",df.shape)
print("Dataset with missing values",df_with_missing.shape)
print("Dataset without missing values",df_without_missing.shape)
```

```python
df_with_missing.head()
```

```python
# import regressor model
from sklearn.ensemble import RandomForestRegressor
# predicta the missing values by random forest imputation
X = df_without_missing.drop(['age'],axis=1)
y = df_without_missing['age']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20, random_state=42)

#using random forest model
rf_model = RandomForestRegressor(n_estimators = 100, random_state=42)
rf_model.fit(X_train,y_train)

#Evaluate the model
y_pred = rf_model.predict(X_test)
print("RMSE",np.sqrt(mean_squared_error(y_test, y_pred)))
print("r2 score",r2_score(y_test, y_pred))
print("MAE",mean_absolute_error(y_test, y_pred))
print("MAPE",mean_absolute_percentage_error(y_test, y_pred))
```

```python
y_pred = rf_model.predict(df_with_missing.drop(['age'], axis=1))
```

```python
print(y_pred)
```

```python
# remove warnings
import warnings
warnings.filterwarnings('ignore')

#replace with predicted values
df_with_missing['age'] = y_pred
df_with_missing.isnull().sum().sort_values(ascending=False)
```
