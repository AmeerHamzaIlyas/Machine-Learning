Data Pre-processing Flow

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
data = sns.load_dataset("titanic")
data.head()
```

```python
#Check null (.isnull()) column values and their sum (.sum()) with proper sorting(.sort())
data.isnull().sum().sort_values(ascending=False)
```

```python
#Drop a column (Deck)
# data = data.drop('deck',axis=1,inplace=True) or 
data.drop('deck',axis=1,inplace=True) # axis = 1 means column and axis =0 means row
# inplace = True means change in original data
```

```python
data.head()
```

1.Imputing values as mode, median, mean etc
for Regression --> mean,median
for Clssification -->mode

```python
# impute missing values with mean/median etc
data['age'] = data['age'].fillna(data['age'].mean())
data.isnull().sum().sort_values(ascending=False)
```

```python
# impute missing values with mode etc
data['embark_town'] = data['embark_town'].fillna(data['embark_town'].mode()[0])
data['embarked'] = data['embarked'].fillna(data['embarked'].mode()[0])
data.isnull().sum().sort_values(ascending=False)
```

2.Imputing values by using algorithm 
 --> K-Nearest Neighbors (KNN)


```python
df = sns.load_dataset("titanic")
df.isnull().sum().sort_values(ascending=False)
```

```python
# Impute Missing values by KNN algorithm
from sklearn.impute import KNNImputer

#Call the knn by the number of neighbours
imputer = KNNImputer(n_neighbors=4)

#impute missing values by KNN imputer
df['age'] = imputer.fit_transform(df[['age']]) # fit means train ans transform means impliment changes by training

#check imputed values
df.isnull().sum().sort_values(ascending=False)
```

Converting Categorical data columns to numerical one using automatical methods like LabelEncoder

```python
#we take all str and category columns for labelen-Coding
from sklearn.preprocessing import LabelEncoder

#columns to encode
columns_to_encode = ['sex', 'embarked', 'who', 'deck', 'class', 'embark_town', 'alive']

# 3dictionary to store labelencoder for each column
label_encoders = {}

#loop to apply labelencoder for the columns
for col in columns_to_encode:
    # call labelencoder object
    le = LabelEncoder()
    # train and transform
    df[col] = le.fit_transform(df[col])
    # add in the dictionary one-one
    label_encoders[col] = le
    
df.head()
```

Imputing the missing values by iterative-imputer


```python
imputer  = IterativeImputer(max_iter=10) #max-iterator means who many times algo check the values by iteration
#columns to impute
columns_to_impute = ['age', 'embark_town', 'embarked', 'deck']

#loop
for col in columns_to_impute:
    df[col] = imputer.fit_transform(df[[col]])
# check missing values
df.isnull().sum().sort_values(ascending=False) 
```
